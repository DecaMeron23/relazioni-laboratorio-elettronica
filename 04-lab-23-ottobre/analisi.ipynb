```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

## Circuiti Digitali

```python
t = np.linspace(0 , 10 , 500)
analogici = 1.5*np.sin(t/3) + 2*np.sin(t)  + np.sin(2*t) + 2 + 0.5 *np.sin(4*t)
media = np.mean(analogici)
digitali = np.where(analogici > media , 5 , 0);

plt.figure(figsize=(14,4))
plt.subplot(121)
plt.title("Segnali Analogici")
plt.plot(t ,analogici)
plt.xlabel("Tempo [s]")
plt.ylabel("Tensione [V]")
plt.subplot(122)
plt.title("Segnali Digitali")
plt.plot(t ,digitali)
plt.xlabel("Tempo [s]")
plt.ylabel("Tensione [V]")
plt.grid(which='both' , linestyle = "--" , linewidth = 0.4)
plt.savefig("immagini/circuiti-digitali/differenza-circuiti-analogici-digitali.png" , dpi = 600 , bbox_inches='tight')
```

## Divisore frequenza

```python
t = np.linspace(0 , 8 , 1000)
v_max = 5
offset = v_max / 2
temp = v_max * np.sin(1*np.pi * t) + offset
temp_1 = v_max * np.sin(0.5*np.pi * t) + offset

clk = np.where(temp > offset , v_max , 0)
D = np.where(temp_1 > offset , v_max , 0)
Q_n = v_max - D
fig , ax = plt.subplots(3 , 1 , figsize = (5,4))
fig.tight_layout(w_pad=0.5)

ax[0].plot(t , clk)
ax[0].set_ylabel("Clock")
ax[0].set_yticks((0 , 5) , (r"$V_{L}$" , r"$V_{H}$") )
ax[0].set_xticks([])
for i in range(0 , 8 , 2):
    ax[0].annotate("", xytext=(i, 2.25), xy=(i, 2.75),
            arrowprops=dict(arrowstyle="->" ,  ec="r"))

ax[1].plot(10*t , D)
ax[1].set_ylabel("Q")
ax[1].set_yticks((0 , 5) , (r"$V_{L}$" , r"$V_{H}$") )

ax[2].plot(10*t , Q_n)
ax[2].set_ylabel(r"$\overline{\mathrm{Q}}$")
ax[2].set_xlabel("Tempo [ms]")
ax[2].set_yticks((0 , 5) , (r"$V_{L}$" , r"$V_{H}$") )

fig.savefig("immagini/flip-flop-D/esempio-divisore-freq.png" , dpi = 600 , bbox_inches='tight')
```
